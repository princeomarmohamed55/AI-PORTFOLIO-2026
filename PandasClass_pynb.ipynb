```python
import pandas as pd
```

**pandas series**
this is a one dimensional structure similar to a column in a table . it is also similar to numpy array

```python
marks = pd.Series([40,56,76,87,43])
print(marks)
```

Series with custom tables

```python
students = pd.Series([80,90,56,44],
                     index =['moksh','mercy', 'rudra','omar']
)
print(students)
```

```python
print(students['omar'])
```
