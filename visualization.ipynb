```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
students =pd.DataFrame({
    "student":["john","mary","ali", "grace",
               "james","faith","brian","lucy"],
    "age":[20,21,22,20,23,21,22,20],
    "gender":["male","female","male","female",
              "male","female","male","female"],
    "math":[78,85,90,66,74,88,82,91],
    "science":[80,79,95,70,88,90,76,64],
    "english":[75,88,92,72,81,86,79,90]
})
```

```python
print(students)
```

```python
#a line chart shows trend over time or ordered data.
plt.plot(students["student"],students["math"])
plt.show()
```

```python
#adding titles
plt.plot(students["student"],students["math"])
plt.title("Math Scores")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.show()
```

```python
#adding titles
plt.plot(students["student"],students["science"])
plt.title("science Scores")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.grid()
plt.show()
```

bar chart

```python
plt.bar(students["student"],students["math"])
plt.show()
```

```python
plt.barh(students["student"],students["math"])
plt.show()
```

```python
plt.barh(students["student"],students["english"])
plt.show()
```

scatter plot:these are used to show relationships between two numerical values


```python
plt.scatter(students["math"],students["science"])
plt.show()
```

Histogram:shows how data is contributed

```python
plt.hist(students["math"])
plt.show()
```

```python
#changing the number of bins
plt.hist(students["math"],bins=5)
plt.show()
```

 **Pie charts:used to show proporpotions**

```python
gender_counts = students["gender"].value_counts()
plt.pie(gender_counts)
plt.show
```

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python

learners = pd.DataFrame({
    'Name': ['Ali', 'Brian', 'Cynthia', 'David', 'Emma', 'Farah', 'George', 'Hassan', 'Ivy', 'John'],
    'Maths': [78, 65, 90, 55, 82, 70, 60, 88, 75, 95],
    'Science': [85, 72, 92, 60, 78, 75, 65, 90, 80, 94],
    'English': [70, 80, 85, 65, 90, 72, 68, 88, 76, 92]
})

learners
```

```python
learners.head(5)
```

```python
plt.plot(learners['Name'], learners['Maths'], marker='o')

plt.title('Maths Scores')
plt.xlabel('Learners')
plt.ylabel('Maths Score')

plt.xticks(rotation=45)
plt.show()
```

```python
plt.bar(learners['Name'], learners['Science'])

plt.title('Science Scores')
plt.xlabel('Learners')
plt.ylabel('Science Score')

plt.xticks(rotation=45)
plt.show()
```

```python
plt.hist(learners['English'], bins=5, edgecolor='black')

plt.title('Distribution of English Scores')
plt.xlabel('English Score')
plt.ylabel('Number of Learners')

plt.show()
```

```python
plt.scatter(learners['Maths'], learners['Science'])

plt.title('Science vs Maths Scores')
plt.xlabel('Maths Score')
plt.ylabel('Science Score')

plt.show()
```

SEARBORN: is a python library bulit on top of matplotlib. it creates beauiful graphics with very little code

```python
import seaborn as sns
```

```python
sns.barplot(data=students,x="gender",y="math")
plt.show()
```

```python
#count plot - count number of observations
sns.countplot(data=students,x="gender")
plt.show()
```

```python
#scatter plot
sns.scatterplot(data=students,x="math",y="science",hue="gender")
plt.show()
```

```python
#heatmap - displays correlations between numerical variables
corr = students[["math","science","english"]].corr()
sns.heatmap(corr,annot=True,cmap="Blues")
plt.show()
```
