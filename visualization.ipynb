```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
```

```python
students =pd.DataFrame({
    "student":["john","mary","ali", "grace",
               "james","faith","brian","lucy"],
    "age":[20,21,22,20,23,21,22,20],
    "gender":["male","female","male","female",
              "male","female","male","female"],
    "math":[78,85,90,66,74,88,82,91],
    "science":[80,79,95,70,88,90,76,64],
    "english":[75,88,92,72,81,86,79,90]
})
```

```python
print(students)
```

```python
#a line chart shows trend over time or ordered data.
plt.plot(students["student"],students["math"])
plt.show()
```

```python
#adding titles
plt.plot(students["student"],students["math"])
plt.title("Math Scores")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.show()
```

```python
#adding titles
plt.plot(students["student"],students["science"])
plt.title("science Scores")
plt.xlabel("Students")
plt.ylabel("Scores")
plt.grid()
plt.show()
```

bar chart

```python
plt.bar(students["student"],students["math"])
plt.show()
```

```python
plt.barh(students["student"],students["math"])
plt.show()
```

```python
plt.barh(students["student"],students["english"])
plt.show()
```

scatter plot:these are used to show relationships between two numerical values


```python
plt.scatter(students["math"],students["science"])
plt.show()
```

Histogram:shows how data is contributed

```python
plt.hist(students["math"])
plt.show()
```

```python
#changing the number of bins
plt.hist(students["math"],bins=3)
plt.show()
```
